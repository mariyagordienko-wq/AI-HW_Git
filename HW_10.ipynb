```python
import random
import statistics
from collections import Counter
```

```python
def generate_random_list(a: int, b: int, length: int) -> list[int]:
    """
    Генерирует список длины length из равномерно распределённых
    случайных целых чисел на отрезке [a, b].

    :param a: нижняя граница диапазона (включительно)
    :param b: верхняя граница диапазона (включительно)
    :param length: длина итогового списка
    :return: список случайных целых чисел
    :raises TypeError: если аргументы не целые числа
    :raises ValueError: если a > b или length отрицательный
    """
    if not all(isinstance(x, int) for x in (a, b, length)):
        raise TypeError("Все аргументы должны быть целыми числами (int)")
    if a > b:
        raise ValueError("Нижняя граница a не может быть больше верхней границы b")
    if length < 0:
        raise ValueError("Длина списка не может быть отрицательной")

    return [random.randint(a, b) for _ in range(length)]


def count_frequencies(values: list) -> dict:
    """
    Строит частотный ряд: считает количество каждого уникального значения.

    :param values: список значений
    :return: словарь {значение: количество}
    """
    # Counter считает вхождения, dict() превращает его в обычный словарь
    return dict(sorted(Counter(values).items()))


def calculate_statistics(values: list) -> dict:
    """
    Считает минимум, среднее, медиану, моду и максимум списка.

    :param values: список чисел
    :return: словарь со статистическими показателями
    :raises ValueError: если список пустой
    """
    if not values:
        raise ValueError("Список пуст — статистики посчитать невозможно")

    return {
        "min": min(values),
        "mean": statistics.mean(values),
        "median": statistics.median(values),
        "mode": statistics.mode(values),
        "max": max(values),
    }


if __name__ == "__main__":
    # Пример из задания
    sample = [1, 2, 3, 1, 2, 2]
    print("Список:", sample)
    print("Частотный ряд:", count_frequencies(sample))
    print("Статистика:", calculate_statistics(sample))

    print("-" * 50)

    # Генерация случайного списка
    random_list = generate_random_list(1, 5, 20)
    print("Случайный список:", random_list)
    print("Частотный ряд:", count_frequencies(random_list))

    stats = calculate_statistics(random_list)
    print("Статистика:")
    for key, value in stats.items():
        print(f"  {key:>7}: {value}")
```
